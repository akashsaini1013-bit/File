# Multithreading
What it is
• Multiple threads run within a single process
• Threads share the same memory space
Key characteristics
• Lightweight and fast to create
• Easy data sharing between threads
• Best for I/O-bound tasks
• In Python (CPython), limited by the Global Interpreter Lock (GIL)
Example use cases
• File I/O
• Network requests
• GUI applications
# Multiprocessing
What it is
• Multiple processes, each with its own memory space
• True parallelism on multiple CPU cores
Key characteristics
• Heavier than threads
• No shared memory by default
• Best for CPU-bound tasks
• Bypasses the Python GIL
Example use cases
• Data processing
• Scientific computing
• Machine learning tasks

```python
# .How can you open a file for writing in Python and write a string to it.
with open('example.txt', 'w') as file:
    file.write('Hello, this is a string written to the file!')
# 2. Write a Python program to read the contents of a file and print each line.
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())
# 3.How would you handle a case where the file doesn’t exist while trying to open it for
# reading
import os
if os.path.exists('example.txt'):
    with open('example.txt', 'r') as file:
        for line in file:
            print(line.strip())
else:
    print("Error: File does not exist.")
# 4.Write a Python script that reads from one file and writes its content to another file
try:
    with open('source.txt', 'r') as source, open('destination.txt', 'w') as dest:
        for line in source:
            dest.write(line)
    print("File copied successfully!")
except FileNotFoundError:
    print("Source file not found.")
except OSError as e:
    print(f"Error: {e}")
# 5.How would you catch and handle division by zero error in Python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return float('inf') # or None, 0, or custom message
print(safe_divide(10, 2)) # 5.0
print(safe_divide(10, 0)) # inf

# 7.How do you log information at different levels (INFO, ERROR, WARNING) in Python
# using the logging module
import logging
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s',
    filename='app.log'
)
logger = logging.getLogger(__name__)

# 8.Write a program to handle a file opening error using exception handling
def read_file_safe(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            return file.read()
    except FileNotFoundError:
        print(f"File '{filename}' not found.")
        return None
    except PermissionError:
        print(f"No permission to read '{filename}'.")
        return None
    except UnicodeDecodeError:
        print(f"Cannot decode '{filename}' with UTF-8 encoding.")
        return None
    except Exception as e:
        print(f"Unexpected error reading '{filename}': {e}")
        return None
# Usage
content = read_file_safe('data.txt')
if content:
    print(content)
# 9.How can you read a file line by line and store its content in a list in Python
try:
    with open('data.txt', 'r', encoding='utf-8') as file:
        lines = [line.strip() for line in file]
    print(f"Read {len(lines)} lines: {lines[:3]}...") # Show first 3
except FileNotFoundError:
    print("File 'data.txt' not found.")
# 10.How can you append data to an existing file in Python
try:
    with open('data.txt', 'a', encoding='utf-8') as file:
        file.write('Appended data with UTF-8 support.\n')
    print("Data appended successfully.")
except PermissionError:
    print("No write permission for file.")
except Exception as e:
    print(f"Error: {e}")
# 11.Write a Python program that uses a try-except block to handle an error when
# attempting to access a Dictionary key that doesn’t exist
my_dict = {'name': 'Alice', 'age': 25}
try:
    role = my_dict['role']
except KeyError:
    print("Key 'role' not found in dictionary.")
    role = 'Unknown'

# 13.How would you check if a file exists before attempting to read it in Python
import os
filename = 'data.txt'
if os.path.exists(filename):
    with open(filename, 'r') as file:
        content = file.read()
    print("File contents:", content)
else:
    print(f"'{filename}' does not exist.")

# 15.Write a Python program that prints the content of a file and handles the case when
# the file is empty
def print_file_lines(filename):
    if not os.path.exists(filename) or os.path.getsize(filename) == 0:
        print(f"File '{filename}' does not exist or is empty.")
        return
    try:
        with open(filename, 'r') as file:
            lines = [line.strip() for line in file if line.strip()]
        if not lines:
            print(f"File '{filename}' contains no non-empty lines.")
        else:
            print(f"Non-empty lines in '{filename}':")
            for i, line in enumerate(lines, 1):
                print(f"{i}: {line}")
    except Exception as e:
        print(f"Error: {e}")
# 16.Demonstrate how to use memory profiling to check the memory usage of a small
# program
import psutil
import os

def get_memory_usage():
    process = psutil.Process(os.getpid())
    return process.memory_info().rss / 1024 / 1024 # MB

initial_memory = get_memory_usage()
print(f"Before: {initial_memory:.2f} MB")

# Memory intensive operation
large_list = [i for i in range(1000000)]

after_list_creation_memory = get_memory_usage()
print(f"After creating list: {after_list_creation_memory:.2f} MB")
print(f"Delta from creation: {after_list_creation_memory - initial_memory:.2f} MB")

del large_list # Free up memory
import gc
gc.collect() # Force garbage collection

after_gc_memory = get_memory_usage()
print(f"After deleting list and GC: {after_gc_memory:.2f} MB")
print(f"Delta after deletion: {after_gc_memory - initial_memory:.2f} MB")

# 17.Write a Python program to create and write a list of numbers to a file, one number per
# lineF
# Create a list of numbers
numbers = [1, 23, 456, 7890, 12345, 67890]
# Write each number on a new line
with open('numbers.txt', 'w') as file:
    for num in numbers:
        file.write(str(num) + '\n')
print("Numbers written to numbers.txt")

# 20.How would you open a file and read its contents using a context manager in Python
def read_file_content(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            content = file.read()
            return content
    except FileNotFoundError:
        return f"File '{filename}' not found"
    except Exception as e:
        return f"Error reading file: {e}"
# Usage
content = read_file_content('example.txt')
print(content)
```
